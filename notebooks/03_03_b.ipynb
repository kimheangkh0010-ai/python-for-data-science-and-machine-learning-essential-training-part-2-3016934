```python
import pandas as pd

import matplotlib.pyplot as plt
from pylab import rcParams
import seaborn as sns

import sklearn
from sklearn.cluster import DBSCAN
from collections import Counter
```

```python
%matplotlib inline
rcParams['figure.figsize'] = 5, 4
sns.set_style("whitegrid")
```

### DBSCan clustering to identify outliers
#### Train your model and identify outliers


```python
df = pd.read_csv(filepath_or_buffer='/Users/kimheang/Documents/Learning 2/python-for-data-science-and-machine-learning-essential-training-part-2-3016934/data/iris.data.csv', header=None, sep=',')

df.columns=['Sepal Length', 'Sepal Width', 'Petal Length', 'Petal Width', 'Species']
data = df.iloc[:, 0:4].values
target = df.iloc[:, 4].values
df[0:5]
```

```python
model = DBSCAN(eps=0.8, min_samples=19).fit(data)
print(model)
```

```python
outliers_df = pd.DataFrame(data)
print(Counter(model.labels_))
print(outliers_df[model.labels_==-1])
```

```python
fig = plt.figure()
ax = fig.add_axes([.1, .1, 1, 1])

color = model.labels_

ax.scatter(data[:,2], data[:,1], c=color, s=120)
ax.set_xlabel('Petal Length')
ax.set_ylabel('Petal Width')
plt.title('DBSCAN for Outliers Detection')
```
